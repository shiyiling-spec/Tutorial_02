4. Create a random numpy array of size 6x3 and transpose it

```python
# 创建一个 6x3 的随机 NumPy 数组
arr2=np.random.rand(6,3)
arr2_T=arr2.T
print(arr2_T)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot produc

```python
n,p,m = 3,4,2

# 创建两个随机 NumPy 数组
arr_A = np.random.rand(n,p)
arr_B = np.random.rand(p,m)

# 计算点积
dot_product = np.dot(arr_A,arr_B)

print("Array A (n x p):\n", arr_A)
print("\nArray B (p x m):\n", arr_B)
print("\nDot Product (n x m):\n", dot_product)
```
